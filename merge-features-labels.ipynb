```python
# 1. Imports
import pandas as pd
```

```python
# 2. Load CSVSs
# Load features and labels
features = pd.read_csv("features.csv")
labels = pd.read_csv("TrainLabels.csv")

print("Features shape:", features.shape)
print("Labels shape:", labels.shape)

# Preview
display(features.head())
display(labels.head())
```

```python
# 3. Merge
# Merge on filename = ClipID
merged = pd.merge(
    features,
    labels,
    left_on="filename",
    right_on="ClipID",
    how="inner"   # only keep matches
)

print("Merged shape:", merged.shape)
display(merged.head())
```

```python
# 4. Reorder Columns
# Use only columns that actually exist
label_cols = [col for col in ["Boredom", "Engagement", "Confusion", "Frustration"] if col in merged.columns]

new_col_order = (
    list(merged.columns[:2]) + 
    label_cols + 
    [col for col in merged.columns[2:] if col not in label_cols and col != "ClipID"]
)

merged = merged[new_col_order]
display(merged.head())
```

```python
# 4. Save Merged CSV
merged.to_csv("merged_features_labels.csv", index=False)
print("✅ Saved merged_features_labels.csv with", len(merged), "rows")
```
